```python
# 匯入套件
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as font_manager

%matplotlib inline
# 設定中文顯示，載入字體: msjh.ttc (微軟正黑體)
font_path = 'msjh.ttc'
font_prop = font_manager.FontProperties(fname=font_path)
plt.rcParams['axes.unicode_minus'] = False    # 避免坐標軸負號顯示問題
```

```python
# 定義 z = y_i * (w^T x_i + b)
z = np.linspace(-2, 3, 400)

# 定義 Hinge Loss 函數
# 當 z >= 1 時，損失為 0；否則為 1 - z
hinge_loss = np.maximum(0, 1 - z)

# 建立圖形
plt.figure(figsize=(8, 5))
plt.plot(z, hinge_loss, color='orange', linewidth=2, label=r'$L(z) = \max(0, 1 - z)$')

# 標示邊界 z=1
plt.axvline(x=1, color='gray', linestyle='--', label='Margin (z=1)')

# 加上標題與座標軸標籤
plt.title("Hinge Loss Function（鉸鏈損失函數）", fontsize=14, fontproperties=font_prop)
plt.xlabel(r"$z = y_i (w^T x_i + b)$", fontsize=12, fontproperties=font_prop)
plt.ylabel("Loss（損失值）", fontsize=12, fontproperties=font_prop)

# 顯示格線與圖例
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.show()
```
