```python
from collections import Counter
doc = nlp(text)
verbs = [token.lemma_ for token in doc if token.pos_ == 'VERB']
# 假設只抓動詞 (VERB)，可以試試看換別的字
verbs_tally = Counter(verbs)
verbs_tally.most_common(10) #前10名
```

```python
import re
from IPython.display import Markdown, display
```

```python
def find_sentences_with_keyword(keyword, document):
    #透過迴圈取得每一句的text
    for sentence in document.sents:
        sentence = sentence.text
        #確定關鍵字有沒有在句子裡(全部變小寫來檢查)
        if keyword.lower() in sentence.lower():
            sentence = re.sub('\n', ' ', sentence) #透過re.sub()將換行符號取代為空白
            sentence = re.sub(f"{keyword}", f"**{keyword}**", sentence, flags=re.IGNORECASE)
            #透過re.sub()將關鍵字取代為加粗字串
            display(Markdown(sentence))
```

```python
find_sentences_with_keyword(keyword="OpenAI", document=doc) # doc = nlp(text)
```
