```python
import spacy
import pandas as pd
nlp = spacy.load('zh_core_web_md')# 對中文語句進行解析
doc = nlp('新冠疫情趨緩，但近來其他疾病如腸病毒、流感等多種傳染病卻呈現流行趨勢')
import pandas as pd
columns=['詞', '詞類', '詞性標注', '單詞依存關係', '是否為純字母組成', '是否為停用詞']
dim = list(map(lambda x: [x.text, x.pos_, x.tag_, x.dep_, x.is_alpha, x.is_stop], doc))
```

```python
pd.DataFrame(dim, columns=columns)
```

```python
from spacy import displacy
```

```python
options = {"compact": True,"bg": "white","color": "black", "distance": 60 }
# doc= nlp('新冠疫情趨緩，但近來其他疾病如腸病毒、流感等多種傳染病卻呈現流行趨勢')
displacy.render(doc, style="dep", options=options)
```

```python
displacy.serve(doc, style="dep")
```

```python
displacy.render(doc, style="ent", options=options)
```

```python
import spacy
nlp = spacy.load('zh_core_web_md')# 對中文語句進行解析
doc = nlp('今天台灣北部的天氣不太好')
for e in doc.ents:
    print(e.text, e.label_)
```

```python
from opencc import OpenCC
```

```python
cc = OpenCC('t2s')
text = '今天台灣北部的天氣不太好'
text = cc.convert(text)
doc = nlp(text)
for e in doc.ents:
    print(e.text, e.label_)
```
